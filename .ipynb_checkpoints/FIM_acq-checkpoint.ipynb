### Calculate start position & XY parameters for exp file(s) according to selected ROI(s)

### this will only run when Magellen has FINISHED acquisition

```python
surface_name = 'New Surface '
surface_n = 1

# Containers to store all ROIs
all_start_pos   = []
all_image_width = []
all_image_height = []
while True:
    try:
        surf = Magellan().get_surface(surface_name+str(surface_n))
        surf.export_to_micro_manager()
        pm = Studio().positions()
        posList = pm.get_position_list()
        x_list=[]
        y_list = []
        for i in range(posList.get_number_of_positions()):
            pos=posList.get_position(i)
            x_list.append([pos.get(0).x])
            y_list.append([pos.get(0).y])
        [start_pos,image_width,image_height] = exp.calculate_ROI(xlist = x_list,ylist = y_list, x_offset = 132, y_offset = -34, x_flip = -1)
        # Save results for this surface
        all_start_pos.append(start_pos)
        all_image_width.append(image_width)
        all_image_height.append(image_height)
        # clear MM poslist after every ROI
        posList.clear_all_positions()
        pm.set_position_list(posList)
        surface_n+=1
    except:
        print('All ROIs exported!')
        break
```

### Return to starting position & generate exp file to start run

```python
exp_name = 'test'
x_pixel = 25
y_pixel = 25

for n in range(0,surface_n-1):
    filename = exp.prepare(exp_name+'_'+str(n),x_pixel,y_pixel, all_image_width[n], all_image_height[n], x_pixel*10)
```

```python
# exp.send_file(filename[0], '146.179.86.142', 5001)
```

## Start Run!

```python
for n in range(0,surface_n-1):
    exp.core.set_xy_position(all_start_pos[n][0],all_start_pos[n][1])
    while exp.core.device_busy('XYStage'):
        # print('getting to starting position')
        continue
    filename = exp.prepare(exp_name+'_'+str(n),x_pixel,y_pixel, all_image_width[n], all_image_height[n], x_pixel*10)
    exp.do_full_raster()
    input("Press Enter to continue to next ROI and/or finish run...")
```
